# QELM and the MIPT: the TFIM reservoir

Starting point: the disordered transverse-field Ising model (TFIM) used as the reservoir. The cell after the autoreload one moves to the repository root.

Notation, the same in every notebook: `N` is the total number of qubits, ancilla included, and `L` the number of samples per run (`samples.L` in the config). Here `N = 6`: six qubits, no ancilla.

```python
import os
from pathlib import Path

# Jupyter starts in the notebook's folder. Move to the repository root (the
# folder that contains pyproject.toml) so conf/, experiments/ and outputs/
# resolve.
root = Path.cwd()
while not (root / "pyproject.toml").is_file() and root != root.parent:
    root = root.parent
os.chdir(root)
print(f"cwd: {Path.cwd()}")
```

```python
from QELM import TransverseFieldIsingModel, load_config, set_seed

cfg = load_config(["sim_type=TFIM", "N=6"])
set_seed(cfg)
model = TransverseFieldIsingModel.from_config(cfg)
print(f"N={model.N} J={model.J} h={model.h} dim={model.dim}")
```

```python
import numpy as np

H = model.hamiltonian().toarray()  # sparse CSR -> dense for the full spectrum
energies = np.linalg.eigvalsh(H)
print(f"H shape={H.shape}, hermitian={np.allclose(H, H.conj().T)}")
print(f"ground-state energy E0={energies[0]:.4f}, gap={energies[1] - energies[0]:.4f}")
```
